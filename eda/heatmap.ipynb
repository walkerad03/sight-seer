```python
from mpl_toolkits.basemap import Basemap
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../raw_images/coords.csv")

lats = df['Latitude'].values
lons = df['Longitude'].values

m = Basemap(
    projection='merc',
    llcrnrlat=23, urcrnrlat=50,
    llcrnrlon=-125, urcrnrlon=-66,
    lat_ts=0,
    resolution='l'
)

x,y = m(lons, lats)

nx_bins = 118
ny_bins = 54

H, xedges, yedges = np.histogram2d(x, y, bins=[nx_bins, ny_bins])
X, Y = np.meshgrid(xedges, yedges)

m.drawcoastlines(linewidth=1, color='black')
m.drawcountries(linewidth=1, color='black')

m.fillcontinents(color='lightgray', lake_color='white')
m.drawmapboundary(fill_color="white")

plt.pcolormesh(X, Y, H.T, cmap='hot', shading='auto', alpha=0.6)
cbar = plt.colorbar()
cbar.set_label('Count')

plt.show()
```
